```python
import pandas as pd
import json
import xml.etree.ElementTree as ET
import io

print("=== 1. CSV Parsing & Missing Values Check ===")
csv_data = """id,name,age,salary
1,Amit,25,50000
2,Rahul,,60000
3,Priya,30,
4,Amit,25,50000
5,Neha,28,75000"""

df_csv = pd.read_csv(io.StringIO(csv_data))
print("Parsed CSV Data:")
print(df_csv)
print("\nMissing Values in CSV:")
print(df_csv.isnull().sum())
print(f"Anomalies (Duplicates found): {df_csv.duplicated().sum()}")

print("\n" + "="*50 + "\n")

print("=== 2. JSON Parsing ===")
json_data = '''
[
  {"id": 101, "product": "Laptop", "price": 75000},
  {"id": 102, "product": "Mouse", "price": null},
  {"id": 103, "product": "Keyboard", "price": 1500}
]
'''
df_json = pd.read_json(json_data)
print("Parsed JSON Data:")
print(df_json)
print("\nMissing Values in JSON:")
print(df_json.isnull().sum())

print("\n" + "="*50 + "\n")

print("=== 3. XML Parsing ===")
xml_data = """
<catalog>
    <book id="1">
        <title>Data Engineering 101</title>
        <price>499</price>
    </book>
    <book id="2">
        <title>Python for Big Data</title>
        <price></price>
    </book>
</catalog>
"""
root = ET.fromstring(xml_data)
xml_list = []
for book in root.findall('book'):
    book_id = book.get('id')
    title = book.find('title').text
    price = book.find('price').text
    xml_list.append({"id": book_id, "title": title, "price": price})

df_xml = pd.DataFrame(xml_list)
print("Parsed XML Data:")
print(df_xml)

print("\n" + "="*50 + "\n")

print("=== 4. HTML Parsing (Table Extraction) ===")
html_data = """
<table>
  <tr><th>Name</th><th>Score</th></tr>
  <tr><td>Vicky</td><td>85</td></tr>
  <tr><td>Rohit</td><td></td></tr>
</table>
"""
df_html_list = pd.read_html(html_data)
df_html = df_html_list[0]
print("Parsed HTML Table Data:")
print(df_html)
print("\nMissing Values in HTML Data:")
print(df_html.isnull().sum())

print("\n" + "="*50 + "\n")

print("=== 5. Text File Parsing ===")
text_data = """
ERROR: Connection timeout at 10:00 AM
INFO: Process started successfully
WARNING: Low memory detected
"""
# Parsing text line by line to extract logs
for line in text_data.strip().split('\n'):
    if "ERROR" in line or "WARNING" in line:
        print(f"Anomaly/Issue Log Found -> {line}")
```
